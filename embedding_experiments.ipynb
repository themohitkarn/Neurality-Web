# Neurality: Collaborative Filtering & Vector Embedding Experiments
This notebook prototypes Collaborative Filtering recommendations using Scikit-Learn Cosine Nearest Neighbors and constructs deep-learning embeddings in PyTorch.

```python
# 1. Scikit-Learn Collaborative Filtering
from sklearn.neighbors import NearestNeighbors
import numpy as np

# Matrix of 50 users x 100 reels
np.random.seed(42)
interaction_matrix = np.random.choice([0.0, 1.0, 5.0, 10.0], size=(50, 100), p=[0.75, 0.15, 0.07, 0.03])

current_user_vector = interaction_matrix[0]

# Fit model
model = NearestNeighbors(metric="cosine", algorithm="brute", n_neighbors=5)
model.fit(interaction_matrix)

distances, indices = model.kneighbors([current_user_vector], n_neighbors=5)
print("Nearest Neighbor distances of closest users:", distances[0])
print("Nearest Neighbor user indices:", indices[0])
```

```python
# 2. PyTorch Deep Learning Embeddings Prototype
import torch
import torch.nn as nn
import torch.optim as optim

class MatrixFactorization(nn.Module):
    def __init__(self, n_users, n_items, emb_dim=16):
        super().__init__()
        self.user_embeddings = nn.Embedding(n_users, emb_dim)
        self.item_embeddings = nn.Embedding(n_items, emb_dim)
        self.user_bias = nn.Embedding(n_users, 1)
        self.item_bias = nn.Embedding(n_items, 1)
        
    def forward(self, user, item):
        dot_prod = (self.user_embeddings(user) * self.item_embeddings(item)).sum(dim=1, keepdim=True)
        return dot_prod + self.user_bias(user) + self.item_bias(item)

# Dummy training sample data
n_users, n_items = 100, 200
mf_model = MatrixFactorization(n_users, n_items, emb_dim=8)
criterion = nn.MSELoss()
optimizer = optim.SGD(mf_model.parameters(), lr=0.1)

# Simulated batch of 32 user interactions
batch_users = torch.randint(0, n_users, (32,))
batch_items = torch.randint(0, n_items, (32,))
batch_ratings = torch.randn(32, 1) # targets

# Forward, backward passes
preds = mf_model(batch_users, batch_items)
loss = criterion(preds, batch_ratings)
optimizer.zero_grad()
loss.backward()
optimizer.step()

print(f"Matrix Factorization Loss on current batch: {loss.item():.4f}")
print("[SUCCESS] Matrix Factorization deep learning prototype successfully built and verified in PyTorch!")
```
